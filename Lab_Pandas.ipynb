```python
personal_df = insurance_df.loc[insurance_df['Policy Type'] == 'Personal Auto']
corporate_df = insurance_df.loc[insurance_df['Policy Type'] == 'Corporate Auto']
```

```python
avg_income_personal = personal_df['Income'].mean()
avg_income_corporate = corporate_df['Income'].mean()
```
