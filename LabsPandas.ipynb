```python
numerical_cols = ['Customer Lifetime Value', 'Income', 'Monthly Premium Auto', 'Total Claim Amount']

print(df[numerical_cols].describe())
```

```python
df['Customer Lifetime Value'] = df['Customer Lifetime Value'].str.replace('%', '')
```

```python
df['Customer Lifetime Value'] = pd.to_numeric(df['Customer Lifetime Value'], errors='coerce')
```

```python
print("Mean:", df['Income'].mean())
print("Median:", df['Income'].median())
print("Mode:", df['Income'].mode().values)
print("Standard Deviation:", df['Income'].std())
print("Quartiles:")
print(df['Income'].quantile([0.25, 0.5, 0.75]))
```

```python
### the mean suggests that the numbers are pulled up by people earning higher wages when analysed with the median

### 25% of customers earn less than $14,072
### 50% earn less than $36,234
### 75% earn less than $64,631
### So, most customers fall under $65k
```

```python
categorical_cols = ['ST', 'GENDER', 'Education', 'Policy Type', 'Vehicle Class']

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(f"Unique values: {df[col].nunique()}")
    print(f"Most frequent value (mode): {df[col].mode().values[0]}")
    print("Value counts:")
    print(df[col].value_counts())
```

```python
policy_counts = df['Policy Type'].value_counts()
print("Policy counts:\n", policy_counts)


top_policy = policy_counts.idxmax()
print("\nMost sold policy type:", top_policy)
```

```python
personal_auto_df = df.loc[df['Policy Type'] == 'Personal Auto']

corporate_auto_df = df.loc[df['Policy Type'] == 'Corporate Auto']

personal_avg_income = personal_auto_df['Income'].mean()
corporate_avg_income = corporate_auto_df['Income'].mean()

print(f"Average Income for Personal Auto customers: ${personal_avg_income:.2f}")
print(f"Average Income for Corporate Auto customers: ${corporate_avg_income:.2f}")
```
