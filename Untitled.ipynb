```python
a = "abc"
b = "deftqs"
mergedString = ""
length_of_strings = min(len(a), len(b))

for i in range (length_of_strings):
    mergedString += a[i] + b[i]
if len(a) < len(b):
    mergedString += b[length_of_strings:]
elif len(b) < len(a):     
    mergedString += a[length_of_strings:]

mergedString

    
```

```python
a = "1 2 3 4 5 6"
b = list(map(int, a.split()))
lowest = b[0]
highest = b[0]


for i in b:
    if i > highest:
        highest = i
    if i < lowest:
        lowest = i
lowest_str=str(lowest) 
highest_str=str(highest)
print(type(lowest_str))
print(type(highest_str))

print(highest_str +" "+ lowest_str)

    
```
