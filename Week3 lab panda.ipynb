```python
# Exercise 3. Compare the average income of customers with Personal Auto and Corporate Auto policies.
```

```python

# Create a DataFrame containing only Personal Auto policies
personal_auto = df.loc[df["Policy Type"] == "Personal Auto"]

# Create a DataFrame containing only Corporate Auto policies
corporate_auto = df.loc[df["Policy Type"] == "Corporate Auto"]

# Calculate the average income for each policy type
personal_auto_income = personal_auto["Income"].mean()
corporate_auto_income = corporate_auto["Income"].mean()

# Print the results
print("Average income for Personal Auto customers:", personal_auto_income)
print("Average income for Corporate Auto customers:", corporate_auto_income)

# Compare the averages
if personal_auto_income < corporate_auto_income:
    print("Customers with Personal Auto policies have a lower average income than customers with Corporate Auto policies.")
elif personal_auto_income > corporate_auto_income:
    print("Customers with Personal Auto policies have a higher average income than customers with Corporate Auto policies.")
else:
    print("Both policy types have the same average income.")
```

```python
# Bonus Exercise 4
#Identify customers whose Total Claim Amount is in the top 25%.
```

```python
#To identify customers with high policy claim amounts, I first calculated the 75th percentile of the Total Claim Amount column using the quantile(0.75) method. Since the top 25% of values are those above the 75th percentile, I filtered the dataset to keep only customers whose Total Claim Amount is greater than this threshold. Finally, I used the describe() method to generate summary statistics for this group of customers and better understand their characteristics.
```

```python

# Calculate the 75th percentile of Total Claim Amount
claim_75 = df["Total Claim Amount"].quantile(0.75)

print("75th percentile of Total Claim Amount:", claim_75)

# Select customers whose Total Claim Amount is above the 75th percentile
high_claim_customers = df[df["Total Claim Amount"] > claim_75]

# Display the DataFrame
print("\nCustomers in the top 25% of Total Claim Amount:")
print(high_claim_customers)

# Summary statistics for high claim customers
print("\nSummary statistics for high claim customers:")
print(high_claim_customers.describe())

# Optional: statistics specifically for Total Claim Amount
print("\nStatistics for Total Claim Amount:")
print(high_claim_customers["Total Claim Amount"].describe())
```
