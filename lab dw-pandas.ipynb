```python
# becauase of the % CLV and number of open complaint are being stored as string(object), we should clean them up and turn them into integers so that we can actually calculate stats like the mean or median
```

```python
file1_df['Customer Lifetime Value'] = file1_df['Customer Lifetime Value'].str.replace('%','').astype(float)

file1_df['Number of Open Complaints'] = file1_df['Number of Open Complaints'].str.extract('(\d+)').astype(float)
```

```python
location_counts = file1_df["ST"].value_counts(ascending=True)
least_common_locations = location_counts.head(5)
least_common_locations
```

```python
policy_counts = file1_df["Policy Type"].value_counts()
print("Policy Frequencies:")
print(policy_counts)
print("-" * 30)
```

```python
highest_policy_type = policy_counts.idxmax()
print(f"The policy type with the highest number of sales is: {highest_policy_type}")
```

```python
personal_auto_df = file1_df.loc[file1_df["Policy Type"] == "Personal Auto"]
corporate_auto_df = file1_df.loc[file1_df["Policy Type"] == "Corporate Auto"]
avg_income_personal = personal_auto_df["Income"].mean()
avg_income_corporate = corporate_auto_df["Income"].mean()
print(f"Average Income for Personal Auto customers: ${avg_income_personal:,.2f}")
print(f"Average Income for Corporate Auto customers: ${avg_income_corporate:,.2f}")
print("=" * 30)
```

```python
claim_75th_percentile = file1_df["Total Claim Amount"].quantile(0.75)
print(f"The 75th_percentile threshold is: ${claim_75th_percentile:,.2f}")
```

```python
high_claim_df = file1_df.loc[file1_df["Total Claim Amount"] >claim_75th_percentile]
high_claim_df["Total Claim Amount"].describe()
```

```python
q75 = file1_df["Total Claim Amount"].quantile(0.75)
print(f"Top 25% Cutoff Threshold:${q75:,.2f}")
print("-" * 30)
high_claim_df = file1_df.loc[file1_df["Total Claim Amount"] > q75]
print("Summary Statistics for High Claim Amount Customers:")
print(high_claim_df["Total Claim Amount"].describe())
```
