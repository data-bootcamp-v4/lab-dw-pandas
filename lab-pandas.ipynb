```python

import pandas as pd

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"

df = pd.read_csv(url)

print("First 5 rows:")
print(df.head())
```

```python

# 1. Dimensions of the dataset
print("Dataset dimensions:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# 2. Data types
print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

# 3. Unique values
print("\nUnique values per column:")
print(df.nunique())

# Categorical columns
categorical_columns = df.select_dtypes(include=["object", "category"])

print("\nCategorical values and frequencies:")
for col in categorical_columns.columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))

# Numerical columns
numerical_columns = df.select_dtypes(include="number")

print("\nNumerical ranges:")
for col in numerical_columns.columns:
    print(f"{col}: Min = {df[col].min()}, Max = {df[col].max()}")

# 4. Numerical summary statistics
print("\nNumerical summary statistics:")
print(numerical_columns.describe())

print("\nMean:")
print(numerical_columns.mean())

print("\nMedian:")
print(numerical_columns.median())

print("\nMode:")
print(numerical_columns.mode().iloc[0])

print("\nStandard deviation:")
print(numerical_columns.std())

print("\nQuartiles:")
print(numerical_columns.quantile([0.25, 0.50, 0.75]))

# 5. Categorical summary statistics
print("\nCategorical summary statistics:")
print(categorical_columns.describe())

# Missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python

# Convert Customer Lifetime Value to numeric
df["Customer Lifetime Value"] = pd.to_numeric(
    df["Customer Lifetime Value"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)

# Convert Number of Open Complaints to numeric
df["Number of Open Complaints"] = pd.to_numeric(
    df["Number of Open Complaints"].astype(str).str.split("/").str[1],
    errors="coerce"
)

# Check updated data types
print(df.dtypes)

# Check numerical statistics
print(df.describe())
```

```python

# Challenge 2 - Exercise 1

location_counts = df["ST"].value_counts()

least_common_locations = location_counts.sort_values().head(5)

print("Top 5 least common customer locations:")
print(least_common_locations)
```

```python

# Challenge 2 - Exercise 2

policy_counts = df["Policy Type"].value_counts()

print("Number of policies by type:")
print(policy_counts)

most_common_policy = policy_counts.idxmax()

print("\nMost common policy type:", most_common_policy)
print("Number of policies:", policy_counts.max())
```

```python

# Challenge 2 - Exercise 3

personal_auto = df.loc[df["Policy Type"] == "Personal Auto"]

corporate_auto = df.loc[df["Policy Type"] == "Corporate Auto"]

personal_income = personal_auto["Income"].mean()

corporate_income = corporate_auto["Income"].mean()

print("Average income - Personal Auto:", personal_income)
print("Average income - Corporate Auto:", corporate_income)

if personal_income < corporate_income:
    print("Personal Auto customers have a lower average income.")
else:
    print("Personal Auto customers have a higher or equal average income.")
```

```python

# Challenge 2 - Bonus Exercise 4

# Calculate the 75th percentile
claim_75 = df["Total Claim Amount"].quantile(0.75)

print("75th percentile:", claim_75)

# Select customers with high claim amounts
high_claim_customers = df.loc[
    df["Total Claim Amount"] > claim_75
]

print("\nNumber of high claim customers:")
print(len(high_claim_customers))

# Summary statistics
print("\nHigh claim customers statistics:")
print(high_claim_customers.describe())

# Display first 5 customers
print("\nFirst 5 high claim customers:")
print(high_claim_customers.head())
```
