## Cargar y visualizar el dataset

```
import pandas as pd

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)

df.head()
```

## Challenge 1: Understanding the data

### 1. Dimensiones del dataset

```
df.shape  # (filas, columnas)
```

### 2. Tipos de datos y correcciones sugeridas

```
df.dtypes
```

### 3. Valores únicos por columna y categóricas

```
unique_counts = df.nunique()
categorical_cols = df.select_dtypes(include='object').columns.tolist()
unique_counts
```

### 4. Estadísticas para columnas numéricas

```
df.describe().T  # Incluye media, std, min, max, percentiles
```

### 5. Estadísticas para columnas categóricas

```
df[categorical_cols].describe(include='object').T
```

## Challenge 2: Analyzing the data

### Ejercicio 1: Top 5 ubicaciones menos comunes

```
state_counts = df['State'].value_counts(ascending=True)
state_counts.head(5)
```

### Ejercicio 2: Número de pólizas por tipo

```
policy_counts = df['Policy Type'].value_counts()
most_common_policy = policy_counts.idxmax()
policy_counts, most_common_policy
```

### Ejercicio 3: Comparación de ingresos entre Personal y Corporate Auto

```
personal = df.loc[df['Policy Type'] == 'Personal Auto']
corporate = df.loc[df['Policy Type'] == 'Corporate Auto']

personal_income = personal['Income'].mean()
corporate_income = corporate['Income'].mean()

personal_income, corporate_income
```

### Bonus - Ejercicio 4: Identificar claims altos (top 25%)

```
# Cálculo del percentil 75
q75 = df['Total Claim Amount'].quantile(0.75)

# Filtrado de clientes con claims altos
high_claims = df[df['Total Claim Amount'] > q75]

# Estadísticas descriptivas
high_claims.describe().T
```
